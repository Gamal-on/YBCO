```python
x, y = tools.tab_interval(cnt.temperature_HPHT, cnt.sample_HC_HPHT, 70, 100)
x1, y1 = tools.tab_interval(cnt.temperature_ISIS, cnt.sample_HC_ISIS, 70, 100)

plt.figure()
plt.plot(x, y, ".r")
plt.plot(x1, y1, ".y")
```

```python
# Plotting the data between 0 and 12 K

squared_temp_HPHT_bounded, hc_HPHT_bounded = tools.tab_interval(cnt.squared_temperature_HPHT, cnt.hc_div_temp_HPHT, 0, 144)
squared_temp_ISIS_bounded, hc_ISIS_bounded = tools.tab_interval(cnt.squared_temperature_ISIS, cnt.hc_div_temp_ISIS, 0, 144)
squared_temp_P24_bounded, hc_P24_bounded = tools.tab_interval(cnt.squared_temperature_P24, cnt.hc_div_temp_P24, 0, 200)


plt.figure()
plt.plot(squared_temp_HPHT_bounded, hc_HPHT_bounded, ".g", label="HPHT")
plt.plot(squared_temp_ISIS_bounded,hc_ISIS_bounded, ".r", label="overdoped_ISIS")
plt.xlim(0,200)
plt.ylim(0, 180)
plt.grid(True)
plt.xlabel("T² (en K²)")
plt.ylabel("C/T (en mJ/K²/mol)")
plt.legend()
plt.legend()
```

```python
# Plotting all the data

plt.figure()
plt.plot(cnt.squared_temperature_HPHT, cnt.hc_div_temp_HPHT, ".g", label="HPHT")
plt.plot(cnt.squared_temperature_ISIS,cnt.hc_div_temp_ISIS, ".r", label="overdoped")
plt.plot(cnt.squared_temperature_P24, cnt.hc_div_temp_P24, "y.", label="P24")
plt.grid(True)
plt.xlabel("Carré de la température (en K²)")
plt.ylabel("C/T (en mJ/K²/mol)")
plt.legend()
```
